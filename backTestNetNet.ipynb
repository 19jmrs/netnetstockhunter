```python
import yfinance as yf
import pandas as pd
```

```python
symbols = ['WETH']

data = yf.Ticker(symbols[0])
```

```python
balancesheet = data.get_balance_sheet(freq="yearly")
```

```python
df = pd.DataFrame(balancesheet)
```

```python
# Assuming your dataframe is named 'df'
df_transposed = balancesheet.reset_index()  # Reset index to make metrics a column
df_transposed.columns.name = None  # Remove column name if any
df_transposed = df_transposed.set_index('index').T  # Set metrics as columns and transpose

# Clean column names (optional)
df_transposed.index = pd.to_datetime(df_transposed.index)  # Convert index to years only
```

```python
df_transposed['TotalLiabilitiesNetMinorityInterest']
```

```python
df_transposed['NCAV'] = df_transposed['CurrentAssets'] - df_transposed['TotalLiabilitiesNetMinorityInterest']
```
